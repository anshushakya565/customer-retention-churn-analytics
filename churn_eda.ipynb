# Let's Understand the Churn Patterns

```python
# Importing all the necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Now load the cleaned dataset
df = pd.read_csv("cleaned_churn_data.csv")
```

## Objective:
Analyze customer churn patterns to identify the key factors contributing to customer attrition and recommend startegies to improve customer retention.

Let's understand
## Churn Distribution

```python
sns.countplot(x = "Churn", data = df)
```

## INSIGHT:
A significant number of customers have churned, indicating opportunites for improving customer retention strategies.

## Churn by Contract Type

```python
sns.countplot(x = "Contract", hue = "Churn", data = df)
```

## INSIGHT:
Customers with month-to-month contracts show significantly higher churn rates compared to long-term contract customers.

## Monthly Charges vs Churn

```python
sns.boxplot(x = "Churn", y = "MonthlyCharges", data = df)
```

## INSIGHT:
Customers with higher monthly charges are more likely to churn.

## Tenure vs Churn

```python
sns.histplot(data = df, x = "tenure", hue = "Churn", multiple= "stack")
```

## INSIGHT:
New customers and customers with shorter tenure show higher churn behavior.

## Internet Service vs Churn

```python
sns.countplot(x = "InternetService", hue = "Churn", data = df)
```

## Tech Support vs Churn

```python
sns.countplot(x = "TechSupport", hue = "Churn", data = df)
```

## INSIGHT:
Customers wihtout technical support services are more likely to churn.

## Correlation Heatmap

```python
#First let us convert categorical values
df_encoded = df.copy()
for col in df_encoded.columns:
    if df_encoded[col].dtype == "object":
        df_encoded[col] = df_encoded[col].astype("category").cat.codes
```

```python
 # Heatmap
plt.figure(figsize = (14,10))
sns.heatmap(df_encoded.corr(), cmap = "Blues")
```

1. Encourage customers to shift toward long-term contracts through discounts and loyalty programs.
